```python
# Question-2
# A dictionary stores student names and marks.
# Write a python program to search for a student's marks using the name entered by the user.
# If the student is not found, display a suitable message.
marks={"Priya":80,"Amit":75,"Rashi":95}
name=input("Enter student name:")
print(marks.get(name,"Student not found"))
```

```python
# Question-3
# A dictionary stores usernames and passwords.
# Write a python program to validate user login.
User={"admin":"1234","student":"2345"}
user_name=input("Username:")
pw=input("Passwaord:")
if User.get(user_name)==pw:
  print("Login Successful")
else:
  print("Invalid Username or Password")
```

```python
# Question-4
# Write a python program to find the student who scored the highest marks.
marks={"Amit":85,"Priya":92,"Jiya":78}
topper=max(marks,key=marks.get)
print("Topper:",topper)
print("Marks:",marks[topper])
```
